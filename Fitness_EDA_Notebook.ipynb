# Fitness Business Data Analysis — Python EDA
Data Analyst Capstone | December 2020

```python
import pandas as pd
import matplotlib.pyplot as plt

INPUT_FILE = 'PowerBI_Dataset.xlsx'
df = pd.read_excel(INPUT_FILE)
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

## 1. Data Quality Check
Check dimensions, data types, missing values, and duplicates.

```python
print('Shape:', df.shape)
print('\nData types:\n', df.dtypes)
print('\nMissing values:\n', df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

## 2. Descriptive Statistics

```python
df.describe().round(2)
```

## 3. Calculated Field Validation
Validate calories per minute as Calories / Duration.

```python
df['Calculated_Cal_per_Min'] = df['Calories'] / df['Duration']
df[['Calories','Duration','Cal_per_Min','Calculated_Cal_per_Min']].head(10)
```

## 4. KPI Summary

```python
kpis = pd.Series({
 'Total Sessions': len(df),
 'Total Calories': df['Calories'].sum(),
 'Avg Calories / Session': df['Calories'].mean(),
 'Avg Calories / Minute': df['Cal_per_Min'].mean(),
 'Avg Duration (min)': df['Duration'].mean(),
 'Avg Pulse': df['Pulse'].mean(),
 'Max Recorded Maxpulse': df['Maxpulse'].max()
})
kpis.round(2)
```

## 5. Intensity Analysis

```python
intensity = df.groupby('Intensity').agg(
 Sessions=('Intensity','size'),
 Total_Calories=('Calories','sum'),
 Avg_Calories=('Calories','mean'),
 Avg_Cal_per_Min=('Cal_per_Min','mean'),
 Avg_Duration=('Duration','mean')
).reindex(['Low','Medium','High'])
intensity.round(2)
```

## 6. Weekday Analysis

```python
weekday = df.groupby('Weekday').agg(
 Sessions=('Weekday','size'),
 Total_Calories=('Calories','sum'),
 Avg_Calories=('Calories','mean')
).round(2)
weekday
```

## 7. Weekly Analysis

```python
weekly = df.groupby('Week').agg(
 Sessions=('Week','size'),
 Total_Calories=('Calories','sum'),
 Avg_Calories=('Calories','mean')
).round(2)
weekly
```

## 8. Visualizations

```python
sorted_df = df.sort_values('Date')
plt.figure(figsize=(10,5))
plt.plot(sorted_df['Date'], sorted_df['Calories'], marker='o')
plt.title('Daily Calories Burned')
plt.xlabel('Date'); plt.ylabel('Calories')
plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

```python
avg = df.groupby('Intensity')['Calories'].mean().reindex(['Low','Medium','High'])
plt.figure(figsize=(7,5)); avg.plot(kind='bar')
plt.title('Average Calories by Intensity')
plt.xlabel('Intensity'); plt.ylabel('Average Calories')
plt.xticks(rotation=0); plt.tight_layout(); plt.show()
```
